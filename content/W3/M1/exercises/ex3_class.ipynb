### Step 1.a Load the data

```python
from sklearn.datasets import load_iris

iris = sklearn.datasets.load_iris(as_frame=True)
df = iris.frame
df.head()
```

### Step 2.a Assign variables `X` to the features and `y` to the target

```python
X = df[iris.feature_names]
y = df["target"]
```

### Step 2.b print the type of each

```python
print(type(X))
print(type(y))
```

### Step 2.c identify whether the target is categorical or numerical & whether the task is regression or classification

```python
print("Target type: categorical")
print("Task type: classification")
```

### Step 3. Identify the number of samples and columns of both the data matrix and the target

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

### Step 4. Summarize the distribution of the data (min, max, median, mean, and standard deviation)

```python
df.describe()
```

### Step 5. Plot the features to identify separability. Can you see clearly distinguishable groups of points?

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.scatter(df["sepal length (cm)"], df["sepal width (cm)"], c=y, cmap="viridis")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Sepal Features")
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df["petal length (cm)"], df["petal width (cm)"], c=y, cmap="viridis")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Petal Features")
plt.show()
```

### Step 6. Keep the following in "Steps 7" onwards

1. First model will trian on the features: `sepal length (cm)` & `sepal width (cm)`
2. Second model will train on the features: `petal length (cm)` & `petal width (cm)`

Finally, evaluate and compare which of the two feature-sets scored better.

Let's do both subsets now..

### Step 6.a subset the features `sepal length (cm)` & `sepal width (cm)`


```python
X_sepal = df[["sepal length (cm)", "sepal width (cm)"]]
```

### Step 6.b subset the features `petal length (cm)` & `petal width (cm)`


```python
X_petal = df[["petal length (cm)", "petal width (cm)"]]
```

### Step 7. Initialize 2 `LogisticRegression` models

```python
from sklearn.linear_model import LogisticRegression

model_sepal = LogisticRegression(max_iter=200)
model_petal = LogisticRegression(max_iter=200)
```

### Step 8. Split the dataset into train and test sets (remember to stratify by the target to account for class imbalance)

```python
from sklearn.model_selection import train_test_split

X_sepal_train, X_sepal_test, y_train, y_test = train_test_split(
    X_sepal, y, test_size=0.2, random_state=42, stratify=y
)

X_petal_train, X_petal_test, _, _ = train_test_split(
    X_petal, y, test_size=0.2, random_state=42, stratify=y
)
```

### Step 9.a Train both models

```python
model_sepal.fit(X_sepal_train, y_train)
model_petal.fit(X_petal_train, y_train)
```

### Step 10. Evaluate and compare both models

```python
acc_sepal = model_sepal.score(X_sepal_test, y_test)
acc_petal = model_petal.score(X_petal_test, y_test)

print("Accuracy using sepal features:", acc_sepal)
print("Accuracy using petal features:", acc_petal)
```

### Step 11. Which of the two subset of features better distinguishes iris flowers? Can you easily tell?

```python
if acc_petal > acc_sepal:
    print("Petal features distinguish iris flowers better.")
    print(f"Petal accuracy: {acc_petal:.3f}  >  Sepal accuracy: {acc_sepal:.3f}")
elif acc_sepal > acc_petal:
    print("Sepal features distinguish iris flowers better.")
    print(f"Sepal accuracy: {acc_sepal:.3f}  >  Petal accuracy: {acc_petal:.3f}")
else:
    print("Both feature sets perform equally well.")
    print(f"Accuracy: {acc_petal:.3f}")
```

### Step 12. Use `DecisionBoundaryDisplay` to plot the learned decision boundary

```python
from sklearn.inspection import DecisionBoundaryDisplay

plt.figure(figsize=(6,4))
DecisionBoundaryDisplay.from_estimator(
    model_petal,
    X_petal,
    response_method="predict",
    cmap="viridis",
    alpha=0.5
)
plt.scatter(X_petal["petal length (cm)"], X_petal["petal width (cm)"], c=y, cmap="viridis")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Decision Boundary (Petal Features)")
plt.show()
```
