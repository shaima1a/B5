# Task 2: Custom conversion

```python
%pip install docling easyocr
```

```python

import json
from pathlib import Path
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.pipeline_options import PdfPipelineOptions, TableStructureOptions
from docling.datamodel.base_models import InputFormat
import os

# PDF
input_doc_path = Path("Shaima.pdf")

# Configure custom conversion
pipeline = PdfPipelineOptions()
pipeline.do_ocr = True
pipeline.ocr_options.lang = ["en"]
pipeline.do_table_structure = True
pipeline.table_structure_options = TableStructureOptions(do_cell_matching=True)

converter = DocumentConverter(
    format_options={InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline)}
)

# Convert
result = converter.convert(input_doc_path)

output_dir = Path("scratch")
output_dir.mkdir(exist_ok=True)

clean_name = "Shaima"

# ---- JSON ----
(output_dir / f"{clean_name}.json").write_text(
    json.dumps(result.document.export_to_dict(), ensure_ascii=False, indent=2),
    encoding="utf-8"
)

# ---- TXT ----
(output_dir / f"{clean_name}.txt").write_text(
    result.document.export_to_markdown(),
    encoding="utf-8"
)

# ---- Markdown ----
(output_dir / f"{clean_name}.md").write_text(
    result.document.export_to_markdown(),
    encoding="utf-8"
)

# ---- Doctags ----
(output_dir / f"{clean_name}.doctags").write_text(
    result.document.export_to_doctags(),
    encoding="utf-8"
)

os.listdir("scratch")
```

# Task 3: Batch conversion

```python
%pip install docling easyocr pyyaml
```

```python
import json
import yaml
import os
from pathlib import Path

from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.datamodel.base_models import InputFormat, ConversionStatus
from docling_core.types.doc import ImageRefMode

# PDF
input_doc_paths = [
    Path("Shaima.pdf"),
    Path("Volunteer Experience.pdf"),
]

pipeline_options = PdfPipelineOptions()
# OCR
pipeline_options.do_ocr = True
# OCR
pipeline_options.ocr_options.lang = ["en"]
# HTML
pipeline_options.generate_page_images = True

doc_converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)
    }
)

# Convert all PDFs
conv_results = doc_converter.convert_all(
    input_doc_paths,
    raises_on_error=False
)

# Export results

output_dir = Path("scratch")
output_dir.mkdir(exist_ok=True)

for conv_res in conv_results:
    if conv_res.status == ConversionStatus.SUCCESS:
        doc = conv_res.document
        name = conv_res.input.file.stem

        # JSON
        (output_dir / f"{name}.json").write_text(
            json.dumps(doc.export_to_dict(), ensure_ascii=False, indent=2),
            encoding="utf-8"
        )

        # HTML
        doc.save_as_html(
            output_dir / f"{name}.html",
            image_mode=ImageRefMode.EMBEDDED
        )

        # Markdown
        (output_dir / f"{name}.md").write_text(
            doc.export_to_markdown(),
            encoding="utf-8"
        )

        # Text
        (output_dir / f"{name}.txt").write_text(
            doc.export_to_markdown(strict_text=True),
            encoding="utf-8"
        )

        # Doctags
        (output_dir / f"{name}.doctags").write_text(
            doc.export_to_doctags(),
            encoding="utf-8"
        )

        # YAML
        (output_dir / f"{name}.yaml").write_text(
            yaml.safe_dump(doc.export_to_dict(), allow_unicode=True),
            encoding="utf-8"
        )
os.listdir("scratch")
```

```python
import os
import json
import yaml
from pathlib import Path

from docling.document_converter import (
    DocumentConverter,
    PdfFormatOption,
)
from docling.datamodel.base_models import InputFormat
from docling.backend.pypdfium2_backend import PyPdfiumDocumentBackend
from docling.pipeline.standard_pdf_pipeline import StandardPdfPipeline

input_paths = [
    Path("Volunteer Experience.pdf"),
    Path("Shaima.pdf"),
    Path("Shaimaa.png"),
]

doc_converter = DocumentConverter(
    allowed_formats=[
        InputFormat.PDF,
        InputFormat.IMAGE,
    ],
    format_options={
        InputFormat.PDF: PdfFormatOption(
            pipeline_cls=StandardPdfPipeline,
            backend=PyPdfiumDocumentBackend,
        ),
    },
)

conv_results = doc_converter.convert_all(input_paths)

output_dir = Path("scratch")
output_dir.mkdir(exist_ok=True)

for res in conv_results:
    if res.document is None:
        print(f"{res.input.file}")
        continue

    name = res.input.file.stem
    doc = res.document

    # Markdown
    (output_dir / f"{name}.md").write_text(
        doc.export_to_markdown(),
        encoding="utf-8"
    )

    # JSON
    (output_dir / f"{name}.json").write_text(
        json.dumps(doc.export_to_dict(), ensure_ascii=False, indent=2),
        encoding="utf-8"
    )

    # YAML
    (output_dir / f"{name}.yaml").write_text(
        yaml.safe_dump(doc.export_to_dict(), allow_unicode=True),
        encoding="utf-8"
    )

os.listdir("scratch")
```

```python

from pathlib import Path
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.datamodel.base_models import InputFormat

#  PDF
input_pdf = Path("Shaima.pdf")

pipeline = PdfPipelineOptions()
pipeline.do_ocr = True                 # OCR
pipeline.ocr_options.lang = ["en"]     # OCR
pipeline.generate_page_images = False

# Create converter
converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline)
    }
)

# Convert the document
result = converter.convert(input_pdf)

# Extract confidence report
confidence = result.confidence

# Print document-level confidence
print("=====================================")
print(" DOCUMENT-LEVEL CONFIDENCE REPORT ")
print(" تقرير الثقة على مستوى المستند ")
print("=====================================")

print("Mean Grade (التقييم العام):", confidence.mean_grade)
print("Low Grade (أسوأ جزء):", confidence.low_grade)

print("\nScores:")
print("Layout Score:", confidence.layout_score)
print("OCR Score:", confidence.ocr_score)
print("Parse Score:", confidence.parse_score)
print("Table Score:", confidence.table_score)

# Page-level confidence
print("\n=====================================")
print(" PAGE-LEVEL CONFIDENCE ")
print(" تقرير الثقة لكل صفحة ")
print("=====================================")

for page_num, page_conf in confidence.pages.items():
    print(f"\n--- Page {page_num} ---")
    print("Layout Score:", page_conf.layout_score)
    print("OCR Score:", page_conf.ocr_score)
    print("Parse Score:", page_conf.parse_score)
    print("Table Score:", page_conf.table_score)

print("\n🎉 Task 5 Completed — Confidence Scores Extracted")
print("تم إنجاز المهمة 5 — تم استخراج درجات الثقة بنجاح")
```
