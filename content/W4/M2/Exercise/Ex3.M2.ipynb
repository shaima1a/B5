## Task 1: Picture Classification

```python
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.datamodel.base_models import InputFormat
```

```python
# PdfPipelineOptions
pipeline_options = PdfPipelineOptions()
pipeline_options.do_picture_classification = True
```

```python
# DocumentConverter
converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)
    }
)
```

```python
result = converter.convert("computerr.pdf")
doc = result.document
```

```python
from docling_core.types.doc import PictureItem

for item in doc.iterate_items():
    if isinstance(item, PictureItem):
        print("Picture ID:", item.id)
        if item.meta and item.meta.classification:
            for pred in item.meta.classification.predictions:
                print("  class:", pred.class_name, "confidence:", pred.confidence)
```

###Integrating Task 1 and Task 2 together in one pipeline

```python
# pipeline options
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.datamodel.base_models import InputFormat

pipeline_options = PdfPipelineOptions()
pipeline_options.do_picture_classification = True
pipeline_options.do_picture_description = True
```

```python
# converter
converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)
    }
)
```

```python
result = converter.convert("computerr.pdf")
doc = result.document
```

```python
from docling_core.types.doc import PictureItem

for item in doc.iterate_items():
    if isinstance(item, PictureItem):
        print("Picture ID:", item.id)

        if item.meta and item.meta.classification:
            for pred in item.meta.classification.predictions:
                print("  class:", pred.class_name, "confidence:", pred.confidence)

        if item.meta and item.meta.description:
            print("  caption:", item.meta.description)
```
